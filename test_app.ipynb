```python
import httpx
import asyncio
```

```python
link = "http://127.0.0.1:8000"
async def main():
    async with httpx.AsyncClient() as client:
        search_response = await client.post(link+'/search', json={'text':['глухой']}, timeout=300)
        print(search_response)
        print(search_response.content.decode("utf-8"))
        print()

        eda_response = await client.post(link+'/eda', json={'text':['глухой']}, timeout=300)
        print(eda_response)
        print(eda_response.content.decode("utf-8"))
        print()
```

```python
await main()
```

```python
import faiss
import torch
dim = 768 # это потом будет лежать в конфиге (когда моделек будет много)
vid_embs = faiss.IndexFlatL2(dim)
vectors = torch.load('./data/embeddings.pt') # и это будет лежать в конфиге
vid_embs.add(torch.mean(vectors, 1).detach())
```

```python
from gensim.models import KeyedVectors
import torch
vectors = torch.load('./data/embeddings.pt') # и это будет лежать в конфиге
dim = 768
index = KeyedVectors(dim)
index.add_vectors([i for i in range(vectors.shape[0])], torch.mean(vectors, 1).detach().numpy())
```

```python
import numpy as np
text_emb = np.random.random((1, 768)).astype('float32')
idxs = index.most_similar(text_emb, topn=2)
print(idxs)
```
